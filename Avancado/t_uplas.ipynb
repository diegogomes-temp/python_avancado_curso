# Tuplas  
Uma tupla em Python é uma estrutura de dados que funciona como uma lista, mas com uma diferença importante: é imutável (não pode ser alterada após criação).

```python
t = 1234, 3456, 3456, 'Hello'
t_lista = [1234,3456,3456,'Hello']
u = t, (1,2,3)
print(t)
print(u)
print(t_lista)
```

```python
x,y,z = t
print(f'elemento x {x}')
print(f'elemento y {y}')
print(f'elemento z {z}')
```

```python
tupla_vazia = ()
tupla_com_um_elemento = 'texto', # Note a vírgula no final para criar uma tupla com um elemento 
print(type(tupla_com_um_elemento))
```

```python
a = t.count(3456)
b = t_lista.count(3456) 
print(f'O elemento 3456 aparece {a} vezes na tupla t')
print(f'O elemento 3456 aparece {b} vezes na lista t_lista')
```
