# Set
Coleção não ordenada de elementos, excluindo os repetidos

```python
cesta = {'maça', 'banana', 'maça', 'uva', 'melancia', 'uva'}
print(cesta) #Observe que elementos duplicados são removidos
```

```python
print('maça' in cesta) # Verifica se 'maça' está no conjunto
```

```python
palavra = set('banana')
print(palavra) # Cada letra única da palavra 'banana' é um elemento do conjunto não ordenado.
```
