# Introdução à Programação Orientada a Objetos

- **Objeto**: algo com dados (atributos) e ações (métodos).
- **Classe**: molde para criar objetos (define quais atributos e métodos existem).
- **Atributos**: informações do objeto (ex.: `nome`, `saldo`).
- **Métodos**: comportamentos do objeto (ex.: `depositar()`, `exibir()`).
- **Por que usar?** Organiza código, evita repetição e facilita manutenção.

Quatro pilares (iremos praticar já): **abstração**, **encapsulamento**, **herança** e **polimorfismo**.

```python
# Exemplo simples de classe e objeto
class ContaBancaria:
    def __init__(self, titular: str, saldo_inicial: float = 0.0):
        self.titular = titular
        self._saldo = saldo_inicial  # underscore indica uso interno

    @property
    def saldo(self) -> float:
        return self._saldo

    def depositar(self, valor: float) -> None:
        if valor <= 0:
            raise ValueError("Valor precisa ser positivo")
        self._saldo += valor

    def sacar(self, valor: float) -> None:
        if valor > self._saldo:
            raise ValueError("Saldo insuficiente")
        self._saldo -= valor

    def resumo(self) -> str:
        return f"Conta de {self.titular} | Saldo: {self._saldo:.2f}"

# Criando objetos e usando métodos
conta = ContaBancaria("Diego", 500)
conta.depositar(250)
conta.sacar(100)
print(conta.resumo())

# Herança e polimorfismo: especializando a conta
class ContaPremium(ContaBancaria):
    def __init__(self, titular: str, saldo_inicial: float = 0.0, cashback: float = 0.02):
        super().__init__(titular, saldo_inicial)
        self.cashback = cashback

    def sacar(self, valor: float) -> None:
        super().sacar(valor)
        self._saldo += valor * self.cashback  # devolve um percentual do saque

conta_vip = ContaPremium("Ana", 1000)
conta_vip.sacar(200)
print(conta_vip.resumo())
```

1) Troque `ContaBancaria` por outro domínio (ex.: `Livro`, `Aluno`, `Produto`).
2) Adicione validações em um método (ex.: `def adicionar_estoque`).
3) Crie uma subclasse que especialize um comportamento (ex.: `ProdutoPerecivel` com aviso de validade).
4) Use `@property` para proteger um atributo e controlar a escrita/leitura.
5) Mostre polimorfismo: duas classes diferentes com o mesmo método `processar()` retornando saídas distintas.

```python
class Pessoa:
    def __init__(self, nome, idade, altura, peso): #construtor
        self.nome = nome
        self.idade = idade
        self.altura = altura
        self.peso = peso
#métodos
    def andar(self):
        print(f"{self.nome} está andando.")

    def falar(self):
        print(f"{self.nome} está falando.")

    def comer(self):
        print(f"{self.nome} está comendo.")

    def imc(self):
        return self.peso / (self.altura ** 2)

    @property #getter
    def altura(self):
        return self._altura

    @altura.setter
    def altura(self, valor):
        self._altura = valor

pessoa1 = Pessoa("Diego", 41, 1.80, 103)
pessoa2 = Pessoa("Ana", 35, 1.65, 68)
# pessoa1.andar()
# pessoa1.falar()
# pessoa1.comer()
print(f"O IMC de {pessoa1.nome} é {pessoa1.imc():.2f}")
print(f"O IMC de {pessoa2.nome} é {pessoa2.imc():.2f}")
print(pessoa1.altura)
```
