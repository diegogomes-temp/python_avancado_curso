## ListComprehension  
Uma sintaxe mais curta para criar uma lista já existente.  
### sintaxe  
`novaLista = [expressão **for item in interable in condition == True**]`


```python
animais = ['cachorro', 'gato', 'papagaio', 'girafa']
novaLista =[]
# for animal in animais:
#     if 'o' in animal:
#         novaLista.append(animal)

novaLista = [animal for animal in animais if 'o' in animal]
print(novaLista)
```

```python
novaLista = [numero for numero in range(10) if numero % 2 == 0]
print(novaLista)
```

```python
animais_MAISCULOS = [animal.upper() for animal in animais]
print(animais_MAISCULOS)
```

```python
novaLista = [animal if animal != 'cachorro' else 'lobo' for animal in animais]
print(f'nova lista com if e else: {novaLista}')
```
