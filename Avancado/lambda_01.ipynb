## Função Lambda
A função **lambda** é uma pequena função anônima que recebe os argumentos porem só pode executar uma expressão.  
`lambda argumento: expressão`

```python
x = lambda a: a*3
print(x(5))
```

```python
soma = lambda a, b: a + b
print(soma(5, 10))
```

```python
graus_celsius = lambda f: (f-32) * 5/9
print(f'A temperatura em Celsius é {graus_celsius(32)}')
```
