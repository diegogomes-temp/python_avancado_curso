# Dicionário  
`dicionario = {chave : valor}`

```python
empregados = {'01': ['João','Silva'],
 '02': ['Maria', 'Souza'],
 '03': ['Pedro', 'Oliveira'],
 '04': ['Ana', 'Maria','Braga']}
print(type(empregados))
```

```python
empregados['04'] # Observe que o dicionário trabalha com chaves e não indices
```

```python
print(empregados['04'][0]) # Acessa o primeiro nome do empregado com chave '04'
print(empregados['04'][1]) # Acessa o sobrenome do empregado com chave '04'
print(empregados['04'][2]) # Acessa o segundo nome do empregado com chave '04'
```

```python
del empregados['03']  # Remove o empregado com chave '03'
print(empregados)
```

```python
empregados['03'] = ['Carlos', 'Pereira']  # Adiciona um novo empregado com chave '03', observe a ordem em que ele é adicionado, no final.
print(empregados)
```

### Alguns métodos 



Os mais usados no dia a dia são: `.get()`, `.keys()`, `.values()`, `.items()`, `.update()` e `.pop()`.

```python
numero_funcionario = input("Digite o número do funcionário: ")
nome = input("Digite o nome do funcionário: ")

if empregados.get(numero_funcionario):
    print("Funcionário já existe.")
else:
    empregados[numero_funcionario] = [nome]
    print("Funcionário adicionado.")

print(empregados)
```

```python
empregados.setdefault('05', ['Luiza', 'Fernandes'])  # Adiciona um novo empregado com chave '05' se não existir
print(empregados)
```

## Ordenar dicionário pela chave

```python
# Ordenar pela chave usando sorted() e dict()
empregados_ordenados = dict(sorted(empregados.items()))
print("Dicionário ordenado pela chave:")
print(empregados_ordenados)
```

```python
empregados_2 = {'06': ['Marcos', 'Antunes'],
 '07': ['Sofia', 'Lima'],
 '08': ['Rafael', 'Costa'],
 '09': ['Beatriz', 'Almeida'],
 '10': ['Lucas', 'Fernandes']}
```

```python
empregados.update(empregados_2)  # Mescla empregados_2 em empregados
print("Dicionário após mesclar empregados_2:")
print(empregados)
```

```python
empregados_ordenados = dict(sorted(empregados.items()))
print("Dicionário ordenado pela chave após mesclar:")
print(empregados_ordenados)
```

```python
dias = {'seg': 'segunda-feira', 'ter': 'terça-feira', 'qua': 'quarta-feira',
        'qui': 'quinta-feira', 'sex': 'sexta-feira', 'sab': 'sábado', 'dom': 'domingo'}
chaves = dias.keys()
print(chaves)
```

```python
for chave in chaves:
    print(f"{chave} -> {dias[chave]}")
```

```python
values = dias.values()
for valor in values:
    print(valor)
```

```python
a = dias.pop('seg')  # Remove o item com chave 'seg' e retorna seu valor
print(f"Dia removido: {a}")
```

```python
dias_copy = dias.copy()  # Cria uma cópia do dicionário dias
print(dias_copy)
```

```python
dias_copy.update({'seg': 'segunda-feira'})  # Adiciona novamente o dia 'seg' na cópia
print(dias_copy)
```
